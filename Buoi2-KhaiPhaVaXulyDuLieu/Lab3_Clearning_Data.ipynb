II – THỰC HÀNH. (Thực hiện trên https://colab.research.google.com)

### 1. Cách lấy dữ liệu dataset từ google drive.

```python
import pandas as pd
import os
import requests

# Cách lấy dữ liệu dataset từ trong google drive
# from google.colab import drive
# drive.mount("/content/drive")

# Tự động tải data.csv từ GitHub nếu chưa có trong môi trường làm việc
if not os.path.exists("data.csv"):
    url = "https://raw.githubusercontent.com/barackvn/MayHocNangCao/main/Buoi2-KhaiPhaVaXulyDuLieu/data.csv"
    r = requests.get(url)
    with open("data.csv", "wb") as f:
        f.write(r.content)

# Đọc file data.csv vào DataFrame
df = pd.read_csv("data.csv")
df.head()
```

---
### 2. Thực hiện các thao tác trên dòng dữ liệu (cột và thuộc tính)

#### 2.1. Thay đổi định dạng chữ hoa / chữ thường trong tên cột

```python
# Thay đổi tiêu đề các dòng thành chữ thường
print(df.columns.str.lower())

### Thay đổi thành chữ in hoa
print(df.columns.str.upper())

### Ký tự đầu in hoa
print(df.columns.str.title())
```

#### 2.2. Thay thế ký tự, khoảng trắng và đổi tên cột

```python
### Thay thế khoảng trắng bằng dấu _
print(df.columns.str.replace(' ', '_'))

### Thay đổi tên cột Age thành Date of Birth
df = df.rename(columns={'Age': 'Date of Birth'})
df.head()
```

#### 2.3. Đổi tên cột theo chỉ số index và lọc cột trừ điều kiện

```python
# Đổi tên cột tại vị trí index 7 thành 'Email Address'
df.columns.values[7] = 'Email Address'
print(df.columns)

### Cho tất cả trừ điều kiện
print(df.columns[df.columns != 'SALARY'])

### Cho tất cả trừ điều kiện dùng phủ định isin
print(df.loc[:, ~df.columns.isin(['SALARY'])].columns)
```

#### 2.4. Chọn cột theo cụm từ, ký tự kết thúc và nhóm tên cột

```python
### Chọn tên cột bắt đầu bằng ký tự trong điều kiện
print(df.filter(like='STREET').columns)

### Chọn tên cột với các ký tự cuối trong điều kiện
print(df.loc[:, df.columns.str.endswith('ame')].columns)

### Hiện tên cột với các ký tự cuối trong điều kiện với bộ lọc và giá trị biến truyền
print(df.filter(regex='ame$', axis=1).columns)

### Hiện một nhóm tên cột
print(df.columns.values[0:4])
```

#### 2.1. Đọc tệp csv vào Dataframe
* Định nghĩa danh sách các định dạng thể hiện giá trị thiếu: `['n/a', 'na', '--']` để pandas nhận diện chuẩn xác.

```python
# Định nghĩa các ký hiệu bị thiếu
missing_values = ["n/a", "na", "--"]

# Đọc tệp csv vào DataFrame
df_prop = pd.read_csv("property_data.csv", na_values=missing_values)

# Hiển thị dữ liệu
df_prop
```

#### 2.2. Tìm trong Dataframe những dữ liệu bị thiếu

```python
# Tìm các vị trí bị thiếu trong toàn bộ DataFrame (True biểu thị dữ liệu bị thiếu)
print(df_prop.isnull())
```

#### 2.3. Xem cột NUM_BEDROOM và tìm những giá trị NaN

```python
# Kiểm tra giá trị NaN trong cột NUM_BEDROOMS
print("Kiểm tra NaN trong cột NUM_BEDROOMS:")
print(df_prop['NUM_BEDROOMS'].isnull())

print("\nCác dòng có NUM_BEDROOMS bị thiếu (NaN):")
print(df_prop[df_prop['NUM_BEDROOMS'].isnull()])
```

#### 2.4. Tổng giá trị còn thiếu trong Dataframe

```python
# Tổng số giá trị bị thiếu theo từng cột
print("Số giá trị bị thiếu theo từng cột:")
print(df_prop.isnull().sum())

# Tổng số giá trị bị thiếu trong toàn bộ Dataframe
total_missing = df_prop.isnull().sum().sum()
print(f"\nTổng cộng số giá trị còn thiếu trong toàn bộ Dataframe: {total_missing}")
```

#### 2.5. Điền giá trị 126 vào ô bị thiếu trong cột “ST_NUM”

```python
# Điền giá trị 126 vào ô bị thiếu trong cột ST_NUM
df_prop['ST_NUM'] = df_prop['ST_NUM'].fillna(126)

print("Cột ST_NUM sau khi điền giá trị 126 vào các ô thiếu:")
print(df_prop[['PID', 'ST_NUM', 'ST_NAME']])
```

#### 2.6. Thay thế missing values trong cột “NUM_BEDROOMS” bằng giá trị trung vị (median)

```python
# Tính giá trị trung vị (median) của cột NUM_BEDROOMS
median_bedrooms = df_prop['NUM_BEDROOMS'].median()
print(f"Giá trị trung vị (median) của cột NUM_BEDROOMS là: {median_bedrooms}")

# Thay thế các giá trị NaN bằng median
df_prop['NUM_BEDROOMS'] = df_prop['NUM_BEDROOMS'].fillna(median_bedrooms)

print("\nBảng dữ liệu sau khi thay thế missing values bằng median:")
print(df_prop)
```
